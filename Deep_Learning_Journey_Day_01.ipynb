Forward Propagation

```python
import numpy as np

def relu(z):
    return np.maximum(0, z)

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# Matches the worked example exactly
x = np.array([[1],[2]])

W1 = np.array([[0.5,-0.3],[0.2,0.8]])
b1 = np.array([[0.1],[-0.1]])

W2 = np.array([[0.6,-0.4]])
b2 = np.array([[0.2]])

# Layer 1
z1 = W1 @ x + b1
a1 = relu(z1)
print("z1:", z1.ravel())    # [0, 1.7]
print("a1:", a1.ravel())    # [0, 1.7]

# Layer 2 (output)
z2 = W2 @ a1 + b2
a2 = sigmoid(z2)
print("z2:", z2.ravel())    # [-0.48]
print("Final prediction (y_hat):", a2.ravel())   # [0.382]
```

```python
import numpy as np

# Sigmoid Activation Function
def sigmoid(z):
    return 1.0 / (1.0 + np.exp(-z))

# Inputs & Targets
X = np.array([0.05, 0.10])
Y = np.array([0.01, 0.99])

# Weights and Biases
W1 = np.array([[0.15, 0.20],
              [0.25, 0.30]])  # Hidden Layer Weights
b1 = 0.35

W2 = np.array([[0.40, 0.45],
              [0.50, 0.55]])  # Output Layer Weights
b2 = 0.60

# --- Vectorized Forward Pass ---
# Layer 1 (Hidden)
net_h = np.dot(W1, X) + b1
out_h = sigmoid(net_h)

# Layer 2 (Output)
net_o = np.dot(W2, out_h) + b2
out_o = sigmoid(net_o)

# Total Error
e_total = np.sum(0.5 * (Y - out_o) ** 2)

print("Hidden Outputs:", np.round(out_h, 7))
print("Output Predictions:", np.round(out_o, 7))
print("Total Error Loss:", round(e_total, 7))
```

Activation Functions

```python
import numpy as np
import matplotlib.pyplot as plt

def sigmoid(z): return 1/(1+np.exp(-z))
def sigmoid_derivative(z): return sigmoid(z)*(1-sigmoid(z))

def tanh(z): return np.tanh(z)
def tanh_derivative(z): return 1 - np.tanh(z)**2

def relu(z): return np.maximum(0, z)
def relu_derivative(z): return (z > 0).astype(float)

def leaky_relu(z, alpha=0.01): return np.where(z>0, z, alpha*z)
def leaky_relu_derivative(z, alpha=0.01): return np.where(z>0, 1, alpha)

def softmax(z):
    exp_z = np.exp(z - np.max(z))  # subtract max for numerical stability
    return exp_z / np.sum(exp_z)

# Verify worked example
print("Sigmoid(2):", sigmoid(2))       # 0.881
print("Tanh(2):", tanh(2))             # 0.964
print("ReLU(2):", relu(2))             # 2
print("LeakyReLU(-3):", leaky_relu(-3))  # -0.03

# Softmax example
logits = np.array([2.0, 1.0, 0.1])
probs = softmax(logits)
print("Softmax probabilities:", probs)  # sums to 1
print("Sum check:", probs.sum())        # 1.0

# Plot all activations
z = np.linspace(-5, 5, 100)
plt.figure(figsize=(10,6))
plt.plot(z, sigmoid(z), label='Sigmoid')
plt.plot(z, tanh(z), label='Tanh')
plt.plot(z, relu(z), label='ReLU')
plt.plot(z, leaky_relu(z), label='Leaky ReLU')
plt.legend(); plt.grid(True)
plt.title('Activation Functions')
plt.show()
```
