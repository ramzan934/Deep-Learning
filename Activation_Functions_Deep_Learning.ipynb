```python

import tensorflow as tf
import numpy as np

# Input values
x = tf.constant([-3.0, -2.0, -1.0, 0.0, 1.0, 2.0, 3.0])

# 1. Sigmoid
sigmoid = tf.keras.activations.sigmoid(x)

# 2. Tanh
tanh = tf.keras.activations.tanh(x)

# 3. ReLU
relu = tf.keras.activations.relu(x)

# 4. Leaky ReLU
leaky_relu = tf.nn.leaky_relu(x, alpha=0.01)

# 5. Softmax
softmax = tf.keras.activations.softmax(x)

# 6. ELU
elu = tf.keras.activations.elu(x)

# Display results
print("Input:       ", x.numpy())
print("Sigmoid:     ", sigmoid.numpy())
print("Tanh:        ", tanh.numpy())
print("ReLU:        ", relu.numpy())
print("Leaky ReLU:  ", leaky_relu.numpy())
print("Softmax:     ", softmax.numpy())
print("ELU:         ", elu.numpy())
```
