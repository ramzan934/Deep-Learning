TASK 01

```python

P = True   # Student is registered for the course
Q = True   # Course fee is paid
R = False  # Active internet connection

# Logical Evaluation: (P AND Q) OR R
can_access = (P and Q) or R

print("=== QUESTION 1: ONLINE COURSE ACCESS ===")
print(f"Inputs -> P={P}, Q={Q}, R={R}")
if can_access:
    print("Decision: ACCESS GRANTED - Student can access the course material.")
else:
    print("Decision: ACCESS DENIED - Student cannot access the course material.")
```

Task 08

```python

print(f"{'P (Red)':<8} | {'Q (Pedestrian)':<15} | {'R (Emergency)':<13} | {'Vehicle Action':<25}")
print("-" * 68)

for P in [True, False]:
    for Q in [True, False]:
        for R in [True, False]:
            if R:
                action = "EMERGENCY VEHICLE PROCEEDS"
            elif P or Q:
                action = "STOP VEHICLE"
            else:
                action = "PROCEED NORMALLY"

            print(f"{str(P):<8} | {str(Q):<15} | {str(R):<13} | {action:<25}")
```
