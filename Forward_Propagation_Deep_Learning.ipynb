```python
import numpy as np
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense

# 1. Input Data Prepare Karein (3 Features)
# Sample shape: (1 batch, 3 features)
X = np.array([[0.5, 1.2, -0.8]], dtype=np.float32)

# 2. Sequential Model Create Karein
model = Sequential([
    # Hidden Layer: 2 Neurons, ReLU activation, Input Shape = 3
    Dense(units=2, activation='relu', input_shape=(3,), name='Hidden_Layer'),

    # Output Layer: 1 Neuron, Sigmoid activation
    Dense(units=1, activation='sigmoid', name='Output_Layer')
])

# 3. Model Architecture Summary Dekhein
model.summary()

# 4. Forward Pass (Prediction) Perform Karein
y_hat = model(X) # ya model.predict(X)


print(f"Input Shape        : {X.shape}")
print(f"Prediction (y_hat) : {y_hat.numpy()[0][0]:.4f}")
```
